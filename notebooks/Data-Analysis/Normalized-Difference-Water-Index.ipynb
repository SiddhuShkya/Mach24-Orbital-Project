### 💧  Normalized Difference Water Index (NDWI)

> The NDWI is used to identify and monitor water bodies, where higher values indicate the presence of water and lower values indicate land or vegetation.

```python
NDWI_IMAGE_PARENT = "../../images/NDWI_PLOTS"
MEAN_OVER_TIME = os.path.join(NDWI_IMAGE_PARENT, "mot.png")
MULTI_YEAR_PLOT = os.path.join(NDWI_IMAGE_PARENT, "ndwi_over_years.png")
SEASONAL_PATTERN_FILE = os.path.join(NDWI_IMAGE_PARENT, "seasonal_pattern.png")
ANNUAL_MEAN_NDWI = os.path.join(NDWI_IMAGE_PARENT, "annual_mean_ndWi.png")
```

### NDWI (Mean Over Time)

```python
img = imread(MEAN_OVER_TIME)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.title("NDWI Mean over Time (Existing Image)")
plt.show()
```

- **`Cyclical Moisture Pattern:`** NDWI shows strong annual seasonality, peaking in autumn (highest water content) and bottoming out in late winter/early spring (driest).

- **`Stable Extremes:`** Peak mean NDWI remains consistent at around 0.13. Trough values are consistently low at approximately 0.03.

- **`High Spatial Variability:`** The large shaded area indicates high uncertainty in water content across the area, particularly during the peak season.

- **`Dry Season Risk:`** The variability band dips into negative values during troughs, suggesting that some areas experience extremely low or no surface moisture at the driest time of year.

✅ NDWI peaks in pre-monsoon or wet months, lower in winter. Shows moisture availability trends.

## Normalized Diffferece Water Index (NDWI) Mean Over Years

```python
img = imread(MULTI_YEAR_PLOT)
plt.figure(figsize=(16, 8))
plt.imshow(img)
plt.axis("off")
plt.show()
```

#### Comparison Across Years

- `Overall Trend in Mean:` There is a slight decreasing trend in the annual mean NDWI over the three years: $0.077$ (2022) $\to$ $0.066$ (2023) $\to$ $0.060$ (2024). This suggests, on average, a minor reduction in the water-related signal (potentially surface water extent or moisture content) over the three-year period.

- `Peak NDWI:` 2022 has the highest observed peak NDWI (around $0.18$ to $0.20$ in October), followed by 2023 and 2024 (both peaking around $0.12$ in October).

- `Seasonal Pattern:` All three years exhibit a clear seasonal cycle, with NDWI values generally lowest in the late winter/early spring months (February-March) and highest in the late fall (October-November), though the exact timing and magnitude of the peaks differ.

- `Variability:` All years show the highest variability (widest shaded area) coinciding with the peak NDWI values in the fall, suggesting the most dynamic or varied water-related conditions occur during this high-water period.

## NDWI Seasonal Pattern

```python
img = imread(SEASONAL_PATTERN_FILE)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

#### Seasonal Trend

The data clearly demonstrates a strong bimodal seasonal pattern in NDWI across all three years:

- `Low NDWI Period:` Values are consistently low (0.03 to 0.06) from January to April. This suggests less surface water or lower moisture content during the late winter and early spring months.

- `High NDWI Periods:` 

    - `Mid-Spring/Early Summer Peak:` There is a noticeable increase in NDWI around May/June.
        - 2022 shows a peak of 0.11 in May.
        - 2023 shows a peak of 0.09 in June.
        - 2024 shows a modest rise to 0.05 in May before the data gap.

    - `Autumn Peak:` The values are highest in the October to December period, particularly in October. This is the strongest water-related signal of the year.
        - The maximum NDWI of 0.13 is observed in October for both 2022 and 2023, and 0.12 in 2024.

- `Overall Trend:` The NDWI values in 2024 appear to be generally lower than the corresponding months in 2022 and 2023, particularly in April (0.04 vs 0.06/0.05), November (0.08 vs 0.09/0.10), and December (0.06 vs 0.07).


## Annual Mean NDWI

```python
img = imread(ANNUAL_MEAN_NDWI)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

The data clearly indicates a decreasing trend in the annual mean NDWI over the three-year period:
