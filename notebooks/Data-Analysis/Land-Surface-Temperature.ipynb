## Initializing Saved Image Paths

```python
LST_IMAGE_PARENT = "../../images/LST_PLOTS"
MEAN_OVER_TIME = os.path.join(LST_IMAGE_PARENT, "mot.png")
MULTI_YEAR_PLOT = os.path.join(LST_IMAGE_PARENT, "lst_over_years.png")
SEASONAL_PATTERN_FILE = os.path.join(LST_IMAGE_PARENT, "seasonal_pattern.png")
ANNUAL_MEAN_LST = os.path.join(LST_IMAGE_PARENT, "annual_mean_lst.png")
```

## Land Surface Temperature (LST) Over Time

```python
img = imread(MEAN_OVER_TIME)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

#### 🌡️ Seasonal Variability and 📈 Year-to-Year Temperature Trends

- **`Extreme Seasonality:`** Shows a massive annual temperature swing, peaking in late spring/early summer (up to ≈35 ∘C) and bottoming in winter (≈15∘C).

- **`High Variability:`** The vast shaded area highlights extreme spatial temperature differences (e.g., 20∘C range in summer), indicating a mix of hot, exposed surfaces and cooler areas. 

- **`Warming Trend:`** The 2024 peak (≈35∘C) was significantly hotter than the 2023 peak (≈29∘C).

✅ LST shows a strong seasonal pattern, highest in April–May (pre-monsoon) and lowest in December–January (winter).

## Land Surface Temperature (LST) Mean Over Years

```python
img = imread(MULTI_YEAR_PLOT)
plt.figure(figsize=(16, 8))
plt.imshow(img)
plt.axis("off")
plt.show()
```

> The image displays the Land Surface Temperature (LST) Mean over Time for three consecutive years (2022, 2023, and 2024), clearly illustrating seasonal cycles and annual changes.




#### 📅 Seasonal Cycles and Peak Temperatures

- `Consistent Seasonality:` All three years exhibit the expected seasonal pattern: low temperatures in the winter months (January/December) and high temperatures in the spring/early summer (March to June).

- `Peak Timing and Intensity:` 
    - The highest LST peak across all years was recorded in April 2024 ($\approx 35.68^{\circ}\text{C}$).
    - The peak temperature appears to shift, occurring earlier in the calendar year over time (Late May 2022 $\rightarrow$ Early June 2023 $\rightarrow$ Late April 2024).

- `Increased Peak Variability:` 
    - 2023 shows the largest fluctuation in LST variability, with the widest shaded area occurring near the May/June 2023 measurements. This suggests a higher temperature contrast across different surface types (e.g., urban vs. vegetated) during this specific period.
    - The variability is generally lowest during the coldest months (Jan/Feb and Nov/Dec).

## LST Seasonal Pattern

```python
img = imread(SEASONAL_PATTERN_FILE)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

#### 🎯 Outliers and Key Events

- `Hottest Observation:` The single highest temperature measurement occurred in April 2024 ($\approx 35.68^{\circ}\text{C}$), which is a clear outlier above the peak mean for both 2022 and 2023, supporting the conclusion of a warmer 2024.

- `Coolest Observations:` The coolest months are consistently January and December, with the absolute minimum LST recorded in January 2022 ($\approx 12.68^{\circ}\text{C}$).

## Annual Mean LST

```python
img = imread(ANNUAL_MEAN_LST)
plt.figure(figsize=(14, 6))
plt.imshow(img)
plt.axis("off")
plt.show()
```

#### 📈 Inter-Annual Warming Trend

- `Increasing Annual Mean:` The average LST for the monitored area shows a clear upward trend across the years:
    - 2022 Mean: $\mathbf{21.38^{\circ}\text{C}}$
    - 2023 Mean: $\mathbf{21.94^{\circ}\text{C}}$
    - 2024 Mean: $\mathbf{23.97^{\circ}\text{C}}$

- The mean LST increased by $\mathbf{2.59^{\circ}\text{C}}$ from 2022 to 2024, suggesting a significant warming trend in the observed landscape.
